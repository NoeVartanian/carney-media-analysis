```python
import pandas as pd
df = pd.read_csv("all_articles - FINAL DATASET.csv")
df.head()
```

```python
df["positive/neutral/negative"] = df["positive/neutral/negative"].str.lower().str.strip()
```

```python

summary = (
    df.groupby(["Country", "positive/neutral/negative"])
      .size()
      .reset_index(name="count")
)


total = df.groupby("Country").size().reset_index(name="total")


summary = summary.merge(total, on="Country")


summary["percent"] = (summary["count"] / summary["total"] * 100).round(2)

summary
```

```python
summary = (
    df.groupby(["Political Bias", "positive/neutral/negative"])
      .size()
      .reset_index(name="count")
)


total = df.groupby("Political Bias").size().reset_index(name="total")


summary = summary.merge(total, on="Political Bias")


summary["percent"] = (summary["count"] / summary["total"] * 100).round(2)

summary
```

```python
summary = (
    df.groupby(["source", "positive/neutral/negative"])
      .size()
      .reset_index(name="count")
)


total = df.groupby("source").size().reset_index(name="total")


summary = summary.merge(total, on="source")


summary["percent"] = (summary["count"] / summary["total"] * 100).round(2)

summary
```

```python
sentiment_col = "positive/neutral/negative"
country_col = "Country"
bias_col = "Political Bias"   #changed the columns based on different analysis


df[sentiment_col] = df[sentiment_col].str.lower().str.strip()
df[bias_col] = df[bias_col].str.strip()


summary = (
    df.groupby([country_col, bias_col, sentiment_col])
      .size()
      .reset_index(name="count")
)


totals = (
    df.groupby([country_col, bias_col])
      .size()
      .reset_index(name="total")
)


summary = summary.merge(totals, on=[country_col, bias_col])


summary["percent"] = (summary["count"] / summary["total"] * 100).round(2)

summary
```

```python
sentiment_col = "open coding/categories"
country_col = "Country"
bias_col = "positive/neutral/negative"  # changed column names depnding on analysis

df[sentiment_col] = df[sentiment_col].str.lower().str.strip()
df[bias_col] = df[bias_col].str.strip()


summary = (
    df.groupby([country_col, bias_col, sentiment_col])
    .size()
    .reset_index(name="count")
)


totals = (
    df.groupby([country_col, bias_col])
    .size()
    .reset_index(name="total")
)


summary = summary.merge(totals, on=[country_col, bias_col])


summary["percent"] = (summary["count"] / summary["total"] * 100).round(2)

summary
```

```python
category_col = "open coding/categories"
country_col = "Country"
bias_col = "Political Bias"
sentiment_col = "positive/neutral/negative"


df[category_col] = df[category_col].astype(str).str.strip()
df[country_col] = df[country_col].astype(str).str.strip()
df[bias_col] = df[bias_col].astype(str).str.strip()
df[sentiment_col] = df[sentiment_col].astype(str).str.lower().str.strip()

counts = (
    df.groupby([country_col, bias_col, category_col, sentiment_col])
      .size()
      .reset_index(name="count")
)


totals = (
    df.groupby([country_col, bias_col, category_col])
      .size()
      .reset_index(name="total")
)


merged = counts.merge(
    totals,
    on=[country_col, bias_col, category_col],
    how="left"
)


merged["percent"] = (merged["count"] / merged["total"] * 100).round(2)


pivot = merged.pivot_table(
    index=[country_col, bias_col, category_col],
    columns=sentiment_col,
    values="percent"
).fillna(0)


pivot = pivot[["positive", "neutral", "negative"]]

pivot
```
